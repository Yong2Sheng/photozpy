## Preset Telescopes

You can import preset telescopes instead of assembing them during the analysis.

The preset telescopes include RM and CT from [Southeastern Association for Research in Astronomy (SARA)](https://www.saraobservatory.org/).

```python
from photozpy.telescope import SARA_RM
from datetime import date
```

### Check basic telescope parameters

```python
SARA_RM.gain()
```

```python
SARA_RM.rdnoise()
```

```python
SARA_RM.filters()
```

```python
# You can also print telescope CCD parameters
# in a summary table.
SARA_RM.print_ccd_summary()
```

```python
# You can print the summary of standadard filter names and their alias
# in a table for a quick look.
SARA_RM.print_filterset_summary()
```

### Get CCD values for specific observation

Telescopes often use several different CCDs over their operational lifetimes. The method shown above is intended only as a quick way to inspect all available CCD parameters. To retrieve the appropriate parameters for a specific night, the recommended approach is to use `photozpy.Telescope.ccd_value()`. It will retrive valid CCD for the night and return the request field values.

```python
# get gain for a specific observation date
SARA_RM.ccd_value(
    field_name="gain",
    obs_date=date(2026,7,22),
)
```

```python
# get readout noise for a specific observation date
SARA_RM.ccd_value(
    field_name="rdnoise",
    obs_date=date(2026,7,22),
)
```

### Filter name matching

Different telescopes may use different names, or aliases, for the same filters. To ensure that the pipeline can process data from a wide range of observatories, these filter names must be standardized.

The available aliases depend on the information provided when the `telescope` object is constructed. For the predefined SARA telescopes, aliases have already been configured for each standard filter name.

```python
SARA_RM.resolve_filter_alias("Sloan g'2")
```
