```python
import astropy.units as u

from photozpy.telescope import SARA_RM
from photozpy.sources.filter_magnitude import (
    FilterMagnitude,
    UPPER_LIMIT_ERROR,
)
```

## Define the magnitudes

```python
# Intentionally use aliases and a non-canonical order:
# r, g, i instead of the Telescope order u, g, r, i, z.
filter_names = (
    "SDSS r",
    "Sloan g'2",
    "SDSS i",
)

magnitudes = u.Quantity(
    [
        19.8421,  # SDSS_r'
        20.1342,  # SDSS_g'
        21.3500,  # SDSS_i': upper-limit magnitude
    ],
    u.mag,
)

errors = u.Quantity(
    [
        0.0285,
        0.0321,
        UPPER_LIMIT_ERROR.to_value(u.mag),
    ],
    u.mag,
)

significance = (
    35.20,
    31.40,
    2.10,
)

magnitude_info = FilterMagnitude(
    telescope=SARA_RM,
    filter_names=filter_names,
    magnitudes=magnitudes,
    errors=errors,
    significance=significance,
    magnitude_system="AB",
)
```

## Check the consturcted filter magnitudes

```python
# Input aliases are resolved to canonical filter names. The stored rows follow
# the filter order defined by the Telescope, including filters with missing data.

print(magnitude_info.filter_names)
```

```python
# Display a formatted summary of the filter magnitudes.

print(magnitude_info.to_table())
```

```python
# Use QTable's native display for an aligned, copy-friendly representation.
# This is convenient when selecting and copying only the required filter rows.

magnitude_info.qtable.pprint_all()
```

## Construct the FilterMagnitude from a QTable

```python
restored = FilterMagnitude.from_qtable(
    telescope=SARA_RM,
    table=magnitude_info.qtable,
    magnitude_system="AB",
)

print(restored.to_table())
```
