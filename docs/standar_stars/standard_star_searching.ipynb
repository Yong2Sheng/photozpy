```python
from photozpy import StandardStarCatalog
```

```python
standard_star = StandardStarCatalog(
    catalog_path="SDSS_DECam.h5",
)
```

```python
standard_star.print_metadata()
```

```python
mag_cols = [
    "SDSS_up",
    "SDSS_gp",
    "SDSS_rp",
    "SDSS_ip",
    "SDSS_zp",
    "UKIDSS_LAS_Y",
]

standard_star_table = standard_star.find_standard_stars(
    fits_path="Master_SDSS J083701.85-011324.4_SDSS_g'.fits",
    edge_margin_pix=100,
    mag_limit=20,
    mag_columns=mag_cols,
)
```

```python
standard_star_table
```

```python
standard_star.plot_standard_stars_on_fits(
    fits_path="Master_SDSS J083701.85-011324.4_SDSS_g'.fits",
    df_stars= standard_star_table,
    hdu_index=0,
    marker_size=80.0,
    marker_edgecolor="red",
    marker_facecolor="none",
    marker_linewidth=1.5,
    cmap="gray",
    stretch="sqrt",
    percent=99.0,
    figsize=(8, 8),
    title="Standard Stars in FOV",
)
```
